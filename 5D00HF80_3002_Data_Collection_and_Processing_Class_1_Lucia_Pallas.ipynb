```python
import datetime
import pandas as pd
```

```python
# 1. upload files
df_emp1 = pd.read_csv('/employees1.csv')
df_emp2 = pd.read_csv('/employees_2.csv') # Ajusta a 'employee2.csv' si ese es su nombre exacto
df_dep = pd.read_csv('/department.csv')
```

```python
# 2. employees (union / concat)
df_emp = pd.concat([df_emp1, df_emp2], ignore_index=True)
```

```python
# 3. employees and departments (inner join)
df = pd.merge(df_emp, df_dep, how='inner', left_on='dep', right_on='id')
```

```python
# 4. clean and rename
df = df.rename(columns={'id_x': 'employee_id', 'dname': 'department_name'})
df = df.drop(columns=['id_y'], errors='ignore')
```

```python
# 5. calculate age (Age)
df['bdate'] = pd.to_datetime(df['bdate'])
current_date = datetime.date.today()
df['Age'] = df['bdate'].apply(
    lambda x: current_date.year - x.year - ((current_date.month, current_date.day) < (x.month, x.day))
)
```

```python
# 6. save first result
df.to_csv('employees-departments.csv', index=False)
print("✅ Tarea 1 completada: 'employees-departments.csv' generado correctamente.")
```

```python
# 7. Format date 'bdate' to text (YYYY-MM-DD)
df['bdate'] = pd.to_datetime(df['bdate']).dt.strftime('%Y-%m-%d')
```

```python
#8. delete extra columns
# Omitimos los IDs (id_x, id_y) y el campo 'image'
df_result = df.drop(columns=['id_x', 'id_y', 'image'], errors='ignore')
```

```python
# 9. reorder columns
columns_order = ['fname', 'lname', 'salary', 'bdate', 'email', 'dep', 'phone1', 'phone2', 'dname', 'age']
```

```python
# 10. hice column employee ID and show only index
df_result = df.drop(columns=['employee_id', 'id', 'id_x', 'id_y'], errors='ignore')
```

```python
# show final result
display(df_result)
```

Task 2

```python
# 1. upload tables
df_projects = pd.read_csv('/projects.csv')
df_works_on = pd.read_csv('/works-on.csv')
```

```python
# 2. merge hours and projects
df_project_details = pd.merge(df_works_on, df_projects, how='inner', left_on='pid', right_on='id')
```

```python
# 3. merge with task 1 employees
df_emp_projects = pd.merge(df, df_project_details, how='inner', left_on='employee_id', right_on='eid')
```

```python
# 4. get pm name using 'mid' with 'employee_id'
df_full = pd.merge(
    df_emp_projects,
    df[['employee_id', 'fname', 'lname']],
    how='inner',
    left_on='mid',
    right_on='employee_id',
    suffixes=('', '_pm')
)
```

```python
# 5. select and rename columns
df_task2 = df_full[[
    'employee_id', 'fname', 'lname', 'hours', 'pname', 'fname_pm', 'lname_pm'
]].rename(columns={
    'employee_id': 'Employee ID',
    'fname': 'First Name',
    'lname': 'Last Name',
    'hours': 'Hours Worked',
    'pname': 'Project Name',
    'fname_pm': 'Manager First Name',
    'lname_pm': 'Manager Last Name'
})
```

```python
# 6. Reorder columns
columns_order = [
    'Employee ID',
    'First Name',
    'Last Name',
    'Hours Worked',
    'Project Name',
    'Manager First Name',
    'Manager Last Name'
]
df_task2 = df_task2[columns_order]
```

```python
# 7. save in CSV
df_task2.to_csv('employees-projects.csv', index=False)
print("✅ Tarea 2 completada: 'employees-projects.csv' generado correctamente.")
```

```python
# show final result
display(df_task2)
```

Task 3

```python
# 1. group by project and calculate hours
df_task3 = df_full.groupby('pname').agg(
    avg_hours=('hours', 'mean'),
    count_eid=('employee_id', 'count')
).reset_index()
```

```python
# 2. Reorder columns
df_task3 = df_task3[['pname', 'avg_hours', 'count_eid']]
```

```python
# 3. save in CSV
df_task3.to_csv('project-work.csv', index=False)
print("✅ Tarea 3 completada: 'project-work.csv' generado correctamente.")
```
